## YOLO

- Python을 기반하는 영상인식 라이브러리
- 라즈베리파이 사용 가능
- 라즈베리파이 5기준
    - YOLOv5 : 10~20FPS
    - YOLOv8n : 8~15FPS
    - YOLOv8s : 3~7FPS
- YOLO 사용시 대부분 Jeston Nano라는 NVIDIA 제품 사용

### 객체 탐지
- Object Detection
    - 이미지, 영상(한 프레임 이미지) 위치와 클래스 모두 예측
    - x,y,w,h 영역안에 물체를 감지
    - 자율주행, 로봇, CCTV 분석 등 활용

- 원리
    - 물체후보영역 추출
    - 각 영역 분류
    - 위 원리를 더 세분화, 그리드화 시킨 알고리즘 활용 : YOLO
