### 웹캠 영상처리

````python
# 라이브러리 로드
import cv2
import numpy as np
import time
````

````python
# 웹캠디바이스 연결하는 데 시간소요
cap = cv2.VideoCapture(0)   # 첫번째 웹캡(0)

while True:
    ret, img = cap.read()   # 웹캠 화면을 하나씩 로드, ret(리턴결과), img(실제 이미지)
    if not ret: break

    cv2.imshow('webcam', img)
    if cv2.waitKey(1) == ord('q'): break    # ord('A') = 65, ord('a' = 97) q(113)키를 누르면 반복 종료

cap.release()   # 메모리 정리
cv2.destroyAllWindows()
````

### FPS(초당 프레임 수) 표시
- 대부분 30FPS를 사용

````python
import time

cap = cv2.VideoCapture(0)   # 첫번째 웹캡(0)

prev_time = 0   # 이전 프레임시간

while True:
    ret, img = cap.read() 
    if not ret: break
    
    curr_time = time.time() # 프레임 출력시간
    
    # FPS 계산
    fps = 1/(curr_time - prev_time) if prev_time != 0 else 0
    prev_time = curr_time

    # 출력할 텍스트
    fps_text = f'FPS: {fps}'
    # img 위에 텍스트 붙이기
    # cv2.putText(img, fps_text, (10, 20), fontFace=cv2.FONT_HERSHEY_SIMPLEX, fontScale=1, color=(0, 255, 0), thickness=2)
    cv2.putText(img, fps_text, (10, 20), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

    cv2.imshow('webcam', img)
    if cv2.waitKey(1) == ord('q'): break    # ord('A') = 65, ord('a' = 97) q(113)키를 누르면 반복 종료

cap.release()   # 메모리 정리
cv2.destroyAllWindows()
````

````python
import cv2
import time

cap = cv2.VideoCapture(0)

# 1. 고해상도 사용 시 필수: MJPG 코덱 설정 (렉 줄이기에 가장 효과적)
cap.set(cv2.CAP_PROP_FOURCC, cv2.VideoWriter_fourcc(*'MJPG'))

# 2. 2K 해상도 설정 (2560 x 1440)
WIDTH, HEIGHT = 2560, 1440
cap.set(cv2.CAP_PROP_FRAME_WIDTH, WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, HEIGHT)

# 실제 카메라가 받아들인 해상도 확인
real_w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
real_h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
print(f"현재 설정된 해상도: {real_w}x{real_h}")

prev_time = 0

while True:
    ret, img = cap.read() 
    if not ret: break
    
    # FPS 계산
    curr_time = time.time() 
    dt = curr_time - prev_time
    fps = 1 / dt if dt > 0 else 0
    prev_time = curr_time

    # 정보 표시 (2K에서는 텍스트가 작아 보일 수 있으니 크기 조정 가능)
    cv2.putText(img, f'FPS: {round(fps, 1)}', (20, 50), 
                cv2.FONT_HERSHEY_SIMPLEX, 1.2, (0, 255, 0), 2)
    cv2.putText(img, f'Res: {real_w}x{real_h}', (20, 100), 
                cv2.FONT_HERSHEY_SIMPLEX, 1.2, (0, 255, 0), 2)
    
    # 3. 화면이 너무 크면 모니터 밖으로 나갈 수 있으니 보여줄 때만 리사이즈 (선택사항)
    display_img = cv2.resize(img, (1280, 720))
    cv2.imshow('webcam', display_img)
    # cv2.imshow('webcam', img)

    key = cv2.waitKey(1) & 0xFF
    if key == ord('q'):
        break
    elif key == ord('s'):
        filename = time.strftime('snapshot_%y%m%d%H%M%S.jpg')
        cv2.imwrite(filename, img) # 저장은 원본 2K로 됨
        print(f'{filename} 저장 완료')

cap.release()
cv2.destroyAllWindows()
````

### 얼굴 검출
- OpenCV 내장된 기능
    - haarcascade_frontalface_default.xml
    - haarcascade_eye_tree_eyeglasses.xml
    - haarcascade_eye.xml
    - haarcascade_smile.xml
    - haarcascade_frontalcatface_extended.xml

````python
import cv2
import time

# Haar Cascade 로드
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)

smile_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_smile.xml'
)

# 웹캠 열기
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError('웹캠 오픈 실패!')

# 해상도 설정
WIDTH, HEIGHT = 640, 480

cap.set(cv2.CAP_PROP_FRAME_WIDTH, WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, HEIGHT)

# 실제 해상도 확인
real_w = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
real_h = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)

prev_time = 0
show_info = False

while True:

    ret, img = cap.read()

    if not ret:
        break

    # 좌우반전
    img = cv2.flip(img, 1)

    # 그레이스케일
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # FPS 계산
    curr_time = time.time()

    fps = 1 / (curr_time - prev_time) if prev_time != 0 else 0
    prev_time = curr_time

    # 정보 표시
    if show_info:

        cv2.putText(
            img,
            f'FPS: {int(fps)}',
            (10, 30),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.9,
            (0, 255, 0),
            2
        )

        cv2.putText(
            img,
            f'{int(real_w)}x{int(real_h)}',
            (10, 70),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.9,
            (0, 255, 0),
            2
        )

    # 얼굴 검출
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(100, 100)
    )

    for (x, y, w, h) in faces:

        # 얼굴 영역 추출
        face_gray = gray[y:y+h, x:x+w]

        # 웃음 검출
        smiles = smile_cascade.detectMultiScale(
            face_gray,
            scaleFactor=1.7,
            minNeighbors=20,
            minSize=(25, 25)
        )

        # 웃는 중이면 텍스트 표시
        if len(smiles) > 0:

            cv2.putText(
                img,
                'Smile :)',
                (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 255),
                2
            )

        # =========================
        # 얼굴 모자이크 처리
        # =========================

        face = img[y:y+h, x:x+w]

        # 작게 축소
        small = cv2.resize(face, (20, 20))

        # 다시 확대
        mosaic = cv2.resize(
            small,
            (w, h),
            interpolation=cv2.INTER_NEAREST
        )

        # 원본에 덮어쓰기
        img[y:y+h, x:x+w] = mosaic

    # 화면 출력
    cv2.imshow('webcam', img)

    key = cv2.waitKey(1) & 0xFF

    # 종료
    if key == ord('q'):
        break

    # 스냅샷 저장
    elif key == ord('s'):

        filename = time.strftime(
            'snapshot_%y%m%d_%H%M%S.jpg'
        )

        cv2.imwrite(filename, img)

        print(f'{filename} 저장 완료')

    # TAB 키
    elif key == 9:

        show_info = not show_info

        print(f'정보 표시: {show_info}')

# 종료 처리
cap.release()
cv2.destroyAllWindows()
````

````python
# Haar Cascade 파일 로드
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml') # 얼굴데이터 따로
smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_smile.xml')   # 미소데이터 따로

cap = cv2.VideoCapture(0)

if not cap.isOpened():  # 웹캠 오픈실패라면
    raise RuntimeError('웹캠 오픈 실패!')

# 해상도 설정
WIDTH, HEIGHT = 640,480 # 1280,720
cap.set(cv2.CAP_PROP_FRAME_WIDTH, WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, HEIGHT)

# 적용된 해상값 확인
real_w = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
real_h = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)

prev_time = 0
show_info = False   # 정보표시 여부
while True:
    ret, img = cap.read() 
    if not ret: break
    
    # FPS
    curr_time = time.time() 
    
    fps = 1/(curr_time - prev_time) if prev_time != 0 else 0
    prev_time = curr_time

    fps_text = f'FPS: {fps}'
    res_text = f'{int(real_w)}x{int(real_h)}'

    # if show_info:
    #     fps_text = f'FPS: {int(fps)}'
    #     res_text = f'{int(real_w)}x{int(real_h)}'

    # cv2.putText(img, fps_text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
    # cv2.putText(img, res_text, (10, 70), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
    
    if show_info:
        cv2.putText(img, f'FPS: {int(fps)}', (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)
        cv2.putText(img, f'{int(real_w)}x{int(real_h)}', (10, 70), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

    # image : 검출할 이미지
    # scaleFactor : 이미지를 몇 배씩 줄여가며 탐색할지 지정하는 값, 1.1 -> 10% 줄이기(속도저하, 정확도상승), 1.7 -> 70% 줄이기
    # 얼굴 1.1 ~ 1,3 (10% ~ 30%), 미소 : 
    # minNeighbors : 후보영역을 진짜 객체로 인정하려면 주변에서 몇번이나 겹쳐 검출되어야 하는지 횟수 
    faces = face_cascade.detectMultiScale(image=img, scaleFactor=1.1, minNeighbors=8)

    # 검출된 물체를 표시
    for(x, y, w, h) in faces:    # x, y 좌측상단좌표, 
        cv2.rectangle(img, (x, y),(x + w, y + h),(255, 0, 255), 2)
        
        # 얼굴 영역만 ROI(Region of Interect)
        face_roi = img[y:y+h, x:x+w]

        # 얼굴 내부에서 미소 검출
        smiles = smile_cascade.detectMultiScale(face_roi,1.7,20)

        for(sx, sy, sw, sh) in smiles:
            cv2.rectangle(face_roi, (sx,sy), (sx+sw, sy+sh), (0, 255, 0),2)

    cv2.imshow('webcam', img)

    key = cv2.waitKey(1) & 0xFF    # 키입력 변수 지정
    if key == ord('q'): 
        break
    elif key ==ord('s'):    # 스냅샷 저장
        filename = time.strftime('snapshot_%y%m%d%H%M%S.jpg')   # ./day06/ 밑에 저장됨
        cv2.imwrite(filename,img)
        print(f'{filename} 스냅샷 저장')
    elif key==9:    # Tab키
        show_info = not show_info   # True->False, False->True
        print(f'표시여부 {show_info}')

cap.release()
cv2.destroyAllWindows()
````

````python
import cv2
import webbrowser

# QR 코드 검출기 생성
qr_detector = cv2.QRCodeDetector()

# 웹캠 열기
cap = cv2.VideoCapture(0)

last_data = ''      # 마지막으로 인식된 데이터 저장
frame_count = 0 
det_interval = 10   # 10프레임마다 검사 (0.3초 정도 간격)

while True:
    ret, img = cap.read()
    if not ret: 
        break
    
    frame_count += 1
    data = ''
    points = None

    # 1. 특정 간격마다만 QR 검사 수행 (성능 최적화)
    if frame_count % det_interval == 0:
        data, points, _ = qr_detector.detectAndDecode(img)

        # QR 코드가 검출되었을 때
        if points is not None and data:
            # 2. 새로운 QR 데이터일 때만 웹브라우저 오픈
            if data != last_data:
                if data.startswith('http://') or data.startswith('https://'):
                    webbrowser.open(data)
                    last_data = data  # 최근 연 사이트 저장
                    print(f"사이트 오픈: {data}")
            
            # (옵션) 인식된 영역 그리기
            points = points.astype(int)
            for i in range(len(points[0])):
                pt1 = tuple(points[0][i])
                pt2 = tuple(points[0][(i + 1) % len(points[0])])
                cv2.line(img, pt1, pt2, (0, 255, 0), 2)
            cv2.putText(img, data, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

        # 3. 화면에 QR이 없으면 인식 기록 초기화 (나중에 다시 열 수 있게 함)
        else:
            # QR이 화면에서 사라지면 last_data를 비워서 
            # 나중에 다시 비췄을 때 인식되도록 함
            last_data = ''

    cv2.imshow("QR Code Reader", img)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
````
