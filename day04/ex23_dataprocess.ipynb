````python
# matplotlib 라이브러리 로드
import matplotlib.pyplot as plt
````

````python
x = [1,2,3,4,5]         # x축 데이터
y = [10,20,15,30,25]    # y축 데이터

plt.plot(x, y)  # 라인차트
plt.show()      # 필요없는 차트설명 제외
````

````python
# 리스소인수, 폰트관리자, 리스소 포함
from matplotlib import rcParams, font_manager, rc

font_path = 'C:/Windows/Fonts/NanumGothicCoding.ttf'
font = font_manager.FontProperties(fname=font_path).get_name()
rc('font',family=font)
rcParams['axes.unicode_minus'] = False  # 마이너스 표시가 깨짐 방지
````

#### 제목, 레이블 표시

- 기본적으로 한글 출력 안됨

````python
plt.plot(x,y)
plt.title('점수 추이')  # 차트 제목
plt.xlabel('시간')      # x축 라벨
plt.ylabel('score')     # y축 라벨
plt.show()
````

- 한글 설정

#### 포인트 모양 변경

````python
plt.plot(x,y, marker='p')   # 마커표시 o, *, ^, v, <, >, s, p
plt.title('점수 추이')      # 차트 제목
plt.xlabel('시간')          # x축 라벨
plt.ylabel('score')         # y축 라벨
plt.show()
````

#### 라인, 색상, 포인트
- x,y 좌표 뒤에 문자열로 작성
    - 색상 : b, g, r, c, m, y, k, w
    - 마커 : 위와 동일
    - 라인스타일 : -(line), --(dash line), -.(dash dot), :(dotted line)

````python
plt.plot(x,y, 'c^--')       # 색상, 마커, 라인 순으로 조합
plt.title('점수 추이')      # 차트 제목
plt.xlabel('시간')          # x축 라벨
plt.ylabel('score')         # y축 라벨
plt.show()
````

````python
# 막대그래프
plt.bar(x, y)
plt.title('점수 추이')
plt.xlabel('시간')
plt.ylabel('점수')
plt.show()
````

#### 차트종류
- plot() : 라인차트
- bar() : 막대차트
- barh() : 가로막대그래프, x, y축 값을 변경
- hist() : 히스토그램, 분포표
- scatter() : 산점도, 데이터간의 관계 표현
- pie() : 파이차트, 비율표시

#### 차트 팁
- grid(True) : 팻플롯립만 사용시는 그리드 표시, seaborn에서는 영향없음
- figure(figSize) : 그래프 크기 설정

#### 3D 그래프

````python
# 주피터노트북을 새로 생성한 뒤 첫번째 코드셸에 라이브러리 배치
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import axes3d     # 3D 그래프용 클래스
import seaborn as sns

import numpy as np

# 리스소인수, 폰트관리자, 리스소 포함
from matplotlib import rcParams, font_manager, rc

font_path = 'C:/Windows/Fonts/NanumGothicCoding.ttf'
font = font_manager.FontProperties(fname=font_path).get_name()
rc('font',family=font)
rcParams['axes.unicode_minus'] = False  # 마이너스 표시가 깨짐 방지
````

````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 30]
z = [5, 10, 8, 12]

fig = plt.figure()  # 새로운 차트 선언해서 변수 할당
ax = fig.add_subplot(111, projection='3d')


ax.set_xlabel('x축')
ax.set_ylabel('y축')
ax.set_zlabel('z축')

plt.title('세 축 데이터')
plt.show()
````

````python
# 3D Surface
x = np.linspace(-5, 5, 50)   # -5 ~ 5 사이를 50개 간격으로 쪼갬
y = np.linspace(-5, 5, 50)   # -5 ~ 5 사이를 50개 간격으로 쪼갬

x, y = np.meshgrid(x, y)
z = x**2 + y**2

fig = plt.figure(figsize=(5,5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x,y,z)
# ax.plot_wireframe(x,y,z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
````
