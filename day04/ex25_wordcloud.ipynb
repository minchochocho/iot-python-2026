````python
# wordcloud 라이브러리 코드
from wordcloud import WordCloud
import matplotlib.pyplot as plt
````

````python
text = '''python is my faborite programming langauge, data analysis. python ai ai ai
        deep learning, machine learning, conteporary ai skill'''

wc = WordCloud().generate(text)
````

````python
plt.imshow(wc)
plt.axis('off') # 차트 x, y축 눈금,레이블 의미하는 객체
plt.show()
````

````python
text = '데이터 분석 파이썬 인공지능 인공지능 데이터 파이썬 인공지능 C++ Java 인공지능 C#'

wc = WordCloud(
    font_path='C:/Windows/Fonts/NanumGothicBold.ttf',
    background_color='azure',
    colormap='plasma'   # virdis, plasma, inferno, cool
    ).generate(text)
plt.imshow(wc)
plt.axis('off') # 차트 x, y축 눈금,레이블 의미하는 객체
plt.show()
````
